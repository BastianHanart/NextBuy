### Deroulement de nos pensees :

Après avoir longuement analysé, trié, combiné et nettoyé nos données, nous avons obtenu un CSV propre et prêt à être utilisé : `CSV_Cleaned.csv`.
Attention toutefois : le fichier est très volumineux et ne peut donc pas être ouvert directement dans son intégralité.


Au début, nous voulions faire un modèle d’IA qui réponde à une question que nous avions définie plus haut.

Après quelques réflexions, nous avons décidé de partir sur un modèle multi‑IA qui nous permettrait de répondre à plusieurs questions à la fois et d’obtenir énormément d’informations sur chaque item, car nous avons choisi de travailler sur les items du magasin et non sur les clients en général.

Pour nous guider et nous challenger, nous sommes partis sur la fabrication d’une pipeline. Une pipeline est un outil très puissant, capable d’entraîner des valeurs sur plusieurs modèles et avec plusieurs paramètres, tout en nécessitant peu de modifications. Nous avons donc décidé de créer une pipeline qui testerait deux grandes familles de modèles : la régression et la classification, car ce sont celles qui nous intéressent le plus ici.

Nous avons commencé par préparer toutes nos données, que nous intégrerons plus tard dans notre pipeline. Nous avons essayé de faire du feature engineering, c’est‑à‑dire créer et utiliser de nouvelles colonnes à partir de celles déjà existantes afin d’augmenter la précision et la cohérence de nos données.
C’est sur ce point que nous pouvons faire le plus d’améliorations : les IA sont pour la plupart loin de leur maximum. Bien que nous parlions de comportements humains — donc très difficiles à reproduire et à prédire — nous pouvons toujours obtenir de meilleurs résultats, notamment grâce à la création d’autres colonnes.
Par manque de temps, nous n’avons pas pu aller plus loin, mais voici une petite liste d’éléments qui auraient pu améliorer grandement la précision :

    Popularité d’un item

    Volume d’articles dans un rayon ou une aile spécifique

    Emplacement dans le magasin

    Acheteur régulier ou non

    …

Malgré cela, nos modèles ont en général une précision avoisinant les 40 à 45 %, ce qui n’est pas négligeable.

La vraie difficulté réside dans la cohabitation entre les différents modèles et les différentes données. Nous parlons de plusieurs dizaines de colonnes, avec parfois plusieurs dizaines de millions de données. Nous arrivons vite à des volumes très importants, qu’il faut donc normaliser. C’est ce que nous avons fait avec notre processeur pour rendre toutes les données valables.
L’optimisation de nos modèles, ainsi que la puissance de nos ordinateurs, sont également des éléments à prendre en compte. Pour certains modèles, le manque de puissance ou de temps pour entraîner correctement notre projet fait que certains modèles pourtant idéaux nous sont inaccessibles.

Néanmoins, grâce à notre pipeline, nous pouvons tout de même ajouter en un clic de nouveaux modèles et entraîner nos données dessus.

### Ameliorations futur


Comme indiqué précédemment, nous pouvons améliorer la qualité, la quantité ou même ajouter certaines données afin d’augmenter la précision et notre score.

Une pipeline peut contenir beaucoup d’hyperparamètres, or nous n’en avons utilisé presque aucun. Leur exploitation pourrait pourtant améliorer considérablement l’apprentissage de nos modèles.

Concernant les paramètres d’entraînement, nous avons coupé nos données à 0,3 pour l’entraînement, mais avec notre pipeline nous pourrions facilement intégrer ce réglage dans notre processus d’automatisation.

Nous pourrions également améliorer notre nombre de prédictions afin de couvrir davantage d’informations.
