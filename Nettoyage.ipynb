### combinaison

```python
# On defini les CSV
CSVProduct = pd.read_csv('datasets/products.csv')
CSVDepartement = pd.read_csv('datasets/departments.csv')
CSVAisles = pd.read_csv('datasets/aisles.csv')
CSVOrderProduct = pd.read_csv('datasets/order_products.csv')
CSVOrder = pd.read_csv('datasets/orders.csv')

# on va merge les column pour ne former plus qu'un
DF = pd.merge(CSVOrder, CSVOrderProduct, on='order_id')
DF = pd.merge(CSVProduct, DF, on='product_id')
DF = pd.merge(CSVAisles, DF, on='aisle_id')
DF = pd.merge(CSVDepartement, DF, on='department_id')

# on retire la column `days since prior order` car elle ne contient aucune valeur
# del DF['days_since_prior_order']

# apres les changement nous arrivons a 13692436


DF = DF.dropna()
# apres notre dropna on arrive a 12811680


# on va maintenant supprimer les column qui nous on servit de cle etrangeres
del DF['product_id']
del DF['aisle_id']
del DF['department_id']



print(DF)

DF.to_csv("CSV_Cleaned.csv", index=False)
```

### Remove des intitule PS dans les noms de produits

```python
#Ici nous allons nous occuper de retirer les PS --> (PS#554551) par exemple
lst = []
new = []
for element in DF["product_name"]:
    if '#' in element:
        if element not in lst:
            if '(PS' in element:
                nouveau = re.sub(r"\(.*?\)", '', element)
                if nouveau not in new:
                    new.append(nouveau)



print("Changement effectue avec succes")


```

```python
# nous allons convertir les code html present dans les noms de produt en reel str valide
DF['product_name'] = DF['product_name'].apply(html.unescape)
isEncore = DF[DF['product_name'].str.contains('&#')]

if isEncore.empty:
    print("Tout les `&#` ont ete change avec succes")
else:
    print(f"Pas ok, il reste {len(isEncore)} a changer")



# TEST ->
# Affiche uniquement la colonne 'product_name' pour ces lignes
# print(CSV_Clean.loc[CSV_Clean['product_name'].str.contains('#'), 'product_name']
```
